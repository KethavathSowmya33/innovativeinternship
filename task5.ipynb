```python
# Social media - Post Engagement analizer
def post_engagement_analyzer(likes_list):
  total_likes=0
  for likes in likes_list:
    total_likes+=likes
  if total_likes>=1000:
    status = "Viral Post"
  else:
    status="Normal Engagement"
  print(f"Total Likes:{total_likes}")
  print(f"Post Status:{status}")
likes_data=[200,150,300,400]
post_engagement_analyzer(likes_data)
```

```python
# Healthcare - Medicine stock alert system
def medicine_stock_alert(stock):
  if stock<10:
    status="Low Stock Alert"
  else:
    status="Stock Sufficient"
  print(f"Medicine Stock: {stock}")
  print(f"Status:{status}")
medicine_stock_alert(6)
```

```python
# Agriculture - Rainfall adequacy Checker
def rainfall_checker(rainfall_data, required_level):
  total=0
  for value in rainfall_data:
    total+=value
  average=total/len(rainfall_data)
  if average>=required_level:
    status="Adequate Rainfall"
  else:
    status="Inadequate Rainfall"
  print(f"Average Rainfall:{average}")
  print(f"Rainfal status:{status}")
rainfall_values=[80,70,65,73]
required=70
rainfall_checker(rainfall_values,required)
```

```python
# Social media- duplicate account detection
def duplicate_account_detection(usernames):
  unique_usernames=set(usernames)
  if len(unique_usernames)<len(usernames):
    print("Duplicate Accounts Found:Yes")
  else:
    print("Duplicate Accounts Found: No")
user_list=["alice","bob","charlie","alice","david"]
duplicate_account_detection(user_list)
```

```python
#health care - Appointment eigibility checker
def appointment_eligibility(age):
  if age>=18:
    status="Eligible"
  else:
    status="Not Eligible"
  print(f"Patient Age:{age}")
  print(f"Eligibility Status:{status}")
appointment_eligibility(21)
```

```python
# Agriculture Premium crop price filter
def premium_crop_filter(prices):
  premium_crops=[]
  for price in prices:
    if price >2000:
      premium_crops.append(price)
  print(f"Premium Crops:{premium_crops}")
crop_prices=[1500,2500,1800,3200,2000]
premium_crop_filter(crop_prices)
```

```python
# System monitoring -  Application health
def application_health_checker(error_count):
  if error_count==0:
    status="Healthy"
  elif error_count<=5:
    status="Minor Issues"
  else:
    status="Critical issues"
  print(f"Error Count:{error_count}")
  print(f"System Status:{status}")
application_health_checker(7)
```

```python
#Banking - Daily transaction limit checker
def transaction_limit_checker(amount):
  daily_limit=50000
  if amount<=daily_limit:
    status="Approved"
  else:
    status="Rejected"
  print(f"Transaction Amount:{amount}")
  print(f"Transaction Status:{status}")
transaction_limit_checker(60000)
```

```python
# E-learning - Students attendance eligibilty system
def attendance_checker(attended_classes, total_classes):
  attendance_percentage=(attended_classes/total_classes)*100
  if attendance_percentage>=75:
    status="Eligible"
  else:
    status="Not Eligible"
  print(f"Attendance Percentage:{attendance_percentage:.1f}")
  print(f"Exam Eligibility:{status}")
attendance_checker(40,50)
```
