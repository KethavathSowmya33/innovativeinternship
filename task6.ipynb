```python
# Smart Parking Lot Management system
def parking_lot_manager(capacity,vechicle_logs):
  parked=0
  peak_usage=0
  for log in vechicle_logs:
    if log=="IN":
      parked+=1
    elif log=="OUT" and parked>0:
      parked-=1
    if parked>peak_usage:
      peak_usage=parked
  if parked>capacity:
    status="Capacity Exceeded"
  else:
    status="Available"
  print(f"Currently Parked Vechicles:{parked}")
  print(f"Peak Parking Usage:{peak_usage}")
  print(f"Parking Status :{status}")
capacity=50
logs=["IN","IN","IN","OUT","IN","IN","OUT"]
parking_lot_manager(capacity,logs)
```

```python
#Online food delivery time estimator
def delivery_time_estimator(distance_km,traffic_level,weather):
  base_time=distance_km*5
  if traffic_level.lower()=="high":
    base_time+=15
  elif traffic_level.lower()=="medium":
    base_time+=10
  elif traffic_level.lower()=="low":
    base_time+=5
  if weather.lower()=="rainy":
    base_time+=10
  elif weather.lower()=="stormy":
    base_time+=20
  elif weather.lower()=="clear":
    base_time+=0
  print(f"Estimated Delivery Time:{base_time} minutes")
delivery_time_estimator(8,"High","Rainy")
```

```python
# Movie Theater seat occupancy analyzer
def seat_occupancy_analyzer(total_seats,booked_seats):
  booked_count=len(booked_seats)
  occupancy_percentage=(booked_count/total_seats)*100
  if booked_count==total_seats:
    status="Housefull"
  elif occupancy_percentage>=75:
    status="Almost Full"
  else:
    status="Available"
  print(f"Occupancy:{occupancy_percentage:.1f}%")
  print(f"Show Status:{status}")
total=200
booked=[1]*150
seat_occupancy_analyzer(total,booked)
```

```python
# Cloud server Load classification system
def server_load_classifier(cpu_readings):
  total=0
  for reading in cpu_readings:
    total+=reading
  average=total/len(cpu_readings)
  if average<50:
    status="Normal"
  elif average<80:
    status="Warning"
  else:
    status="Critical"
  print(f"Average CPU Load:{average:.1f}%")
  print(f"Server status:{status}")
cpu_data=[45,60,70,85,90]
server_load_classifier(cpu_data)
```

```python
#Smart classroom resource usage monitor
def resource_usage_monitor(usage_data):
  threshold=8
  overused=[]
  for resource,hours in usage_data.items():
    if hours>threshold:
      overused.append(resource)
  if overused:
    alert="Yes"
  else:
    alert="No"
  print(f"Overused Resources:{','.join(overused) if overused else 'None'}")
  print(f"Energy Alert:{alert}")
usage={
    "Projector":6,
    "AC":9,
    "Lights":4
}
resource_usage_monitor(usage)
```

```python
# Online Event Registartion capacity Controller
def event_registration_manager(capacity,registrations):
  if registrations<=capacity:
    confirmed=registrations
    waitlisted=0
    status="Open"
  else:
    confirmed=capacity
    waitlisted=registrations-capacity
    status="Closed"
  print(f"Confirmed registration:{confirmed}")
  print(f"Waitlisted Users:{waitlisted}")
  print(f"Registration status:{status}")
event_capacity=100
total_registrations=105
event_registration_manager(event_capacity,total_registrations)

```
