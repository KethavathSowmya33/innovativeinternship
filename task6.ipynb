```python
#Smart Parking Lot Management System
def smart_parking_system(capacity,logs):
  parked=0
  peak_usage=0
  for log in logs:
    if log=="IN":
      parked+=1
      if parked>peak_usage:
        peak_usage=parked
    elif log=="OUT":
      if parked>0:
        parked-=1
  if parked>capacity:
    status="Alert:Parking Over Capacity"
  elif parked==capacity:
    status="Parking Full"
  else:
    status="Available"
  return parked,peak_usage,status
capacity=50
vechicle_logs=["IN","IN","IN","OUT","IN","IN","OUT"]
parked,peak,status=smart_parking_system(capacity,vechicle_logs)
print("Currently Parked Vechicles:",parked)
print("Peak Usage:",peak)
print("Parking Status:",status)
```

```python
#Online Food Delivery Time Estimator
def delivery_time_estimator(distance_km,traffic_level,weather):
  base_time=distance_km*5
  if traffic_level.lower()=="low":
    traffic_delay=0
  elif traffic_level.lower()=="medium":
    traffic_delay=10
  elif traffic_level.lower()=="high":
    traffic_delay=20
  else:
    traffic_delay=0
  if weather.lower()=="clear":
    weather_delay=0
  elif weather.lower()=="rainy":
    weather_delay=15
  elif weather.lower()=="stormy":
    weather_delay=25
  else:
    weather_delay=0
  estimated_time=base_time+traffic_delay+weather_delay
  return estimated_time
distance=8
traffic="High"
weather="Rainy"
eta=delivery_time_estimator(distance,traffic,weather)
print("Estimated Delivery Time:",eta,"minutes")

```

```python
#Movie Theater Seat Occupancy Analyzer
def seat_occupancy_analyzer(total_seats,booked_seats):
  booked_count=len(booked_seats)
  occupancy_percentage=(booked_count/total_seats)*100
  if booked_count==total_seats:
    status="Housefull"
  elif occupancy_percentage>=75:
    status="Almost Full"
  elif occupancy_percentage>=50:
    status="Moderate"
  else:
    status="Plenty of seats Available"
  suggestion="Consider opening additional shows" if occupancy_percentage>=90 else "No need for extra shows"
  return occupancy_percentage,status,suggestion
total_seats=200
booked_seats=[1]*150
occupancy,status,suggestion=seat_occupancy_analyzer(total_seats,booked_seats)
print("Occupancy:",f"{occupancy:.2f}%")
print("Show status:",status)
print("Suggestion:",suggestion)
```

```python
#Cloud Server Load Classification System
def server_load_classifier(cpu_readings):
  avg_cpu=sum(cpu_readings)/len(cpu_readings)
  if avg_cpu<50:
    status="Normal"
  elif 50<=avg_cpu<80:
    status="Warning"
  else:
    status="Critical"
  return avg_cpu,status
cpu_readings=[45,60,70,85,90]
avg,status=server_load_classifier(cpu_readings)
print("Average CPU Load:",f"{avg:.2f}%")
print("Server Status:",status)
```

```python
#Smart Classroom Resource Usage Monitor
def classroom_resource_monitor(resource_usage):
  thresholds={
      "Projector":5,
      "AC":8,
      "Lights":6
  }
  overused=[]
  for resource, hours in resource_usage.items():
    if resource in thresholds and hours > thresholds[resource]:
      overused.append(resource)
  energy_alert="Yes" if overused else "No"
  return overused, energy_alert
usage_data={
    "Projector":6,
    "AC":9,
    "Lights":4
}
overused_resources,alert=classroom_resource_monitor(usage_data)
print("Overused resources:",",".join(overused_resources) if overused_resources else "None")
print("Energy Alert:", alert)
```

```python
# Online event registration capacity controller
def event_registration_controller(capacity,registrations):
  if registrations<=capacity:
    confirmed=registrations
    waitlisted=0
    status="Open"
  else:
    confirmed=capacity
    waitlisted=registrations-capacity
    status="Closed"
  return confirmed,waitlisted,status
event_capacity=100
registrations=105
confirmed,waitlisted,status=event_registration_controller(event_capacity,registrations)
print("Confirmed Registrations:",confirmed)
print("Waitlisted Users:",waitlisted)
print("Registration Status:",status)
```
