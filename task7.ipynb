```python
# Smart Light Controller
class SmartLight:
  def __init__(self,name):
    self.name=name
    self.status="OFF"
  def turn_on(self):
    self.status="ON"
  def turn_off(self):
    self.status="OFF"
  def display_status(self):
    print(f"{self.name} is {self.status}")
light=SmartLight("Bedroom Light")
light.turn_on()
light.display_status()
```

```python
#Employee ID card System
class Employee:
  def __init__(self,name,emp_id,department):
    self.name=name
    self.emp_id=emp_id
    self.department=department
  def display_id_card(self):
    print("----Employee ID Card ----")
    print(f"Name      : {self.name}")
    print(f"Employee ID: {self.emp_id}")
    print(f"Department:{self.department}")
    print("------------------------------")
employee1=Employee("Rahul","EMP102", "IT")
employee1.display_id_card()
```

```python
# Mobile Contact Record
class Contact:
  def __init__(self,name,phone):
    self.name=name
    self.phone=phone
  def display_contact(self):
    print("Contact Saved")
    print(f"Name: {self.name}")
    print(f"Phone: {self.phone}")
contact1=Contact("Anita","9876543210")
contact1.display_contact()
```

```python
# Product Price tag Generator
class Product:
  def __init__(self,name,price):
    self.name=name
    self.price=price
  def display_price_tag(self):
    print(f"Product: {self.name}")
    print(f"Price: {self.price}")
product1= Product("Headphones",2499)
product1.display_price_tag()
```

```python
#Movie rating Display System
class Movie:
  def __init__(self,name,rating):
    self.name=name
    self.rating=rating
  def display_details(self):
    print("----Movie Details ----")
    print(f"Movie name : {self.name}")
    print(f"Rating     : {self.rating}")
    print("----------------------------")
movie1=Movie("Inception",4.8)
movie1.display_details()
```

```python
#Delivery Address manager
class Delivery:
  def __init__(self,customer_name,address):
    self.customer_name=customer_name
    self.address=address
  def display_details(self):
    print("Delivery Details")
    print(f"Customer: {self.customer_name}")
    print(f"Address: {self.address}")
delivery1=Delivery("Suman", "Hyderabad")
delivery1.display_details()
```
